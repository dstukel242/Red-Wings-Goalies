```python
import pandas as pd

file_id = '1Wky-flLFRBfnNw0nPiSUdr2oq4ZFN-rI'
url = f'https://drive.google.com/uc?id={file_id}'

df = pd.read_csv(url)
df = df.drop(columns=['Unnamed: 0'])
df = df.rename(columns={'Unnamed: 1': 'Player'})
df = df.rename(columns={'Unnamed: 2': 'From'})
df = df.rename(columns={'Unnamed: 3': 'To'})
df = df.rename(columns={'Unnamed: 4': 'Years Played'})
df = df.rename(columns={'Goalie Stats': 'GP'})
df = df.rename(columns={'Goalie Stats.1': 'GS'})
df = df.rename(columns={'Goalie Stats.2': 'Wins'})
df = df.rename(columns={'Goalie Stats.3': 'Losses'})
df = df.rename(columns={'Goalie Stats.4': 'Ties + OT Losses'})
df = df.rename(columns={'Goalie Stats.5': 'Goals Against'})
df = df.rename(columns={'Goalie Stats.6': 'Shots Against'})
df = df.rename(columns={'Goalie Stats.7': 'Saves'})
df = df.rename(columns={'Goalie Stats.8': 'Save %'})
df = df.rename(columns={'Goalie Stats.9': 'Goals Against Avg'})
df = df.rename(columns={'Goalie Stats.10': 'Shutouts'})
df = df.rename(columns={'Goalie Stats.11': 'Minutes Played'})
df = df.rename(columns={'Goalie Stats.12': 'Quality Starts'})
df = df.rename(columns={'Goalie Stats.13': 'Quality Start %'})
df = df.rename(columns={'Goalie Stats.14': 'Really Bad Starts'})
df = df.rename(columns={'Goalie Stats.15': 'GA % -'})
df = df.rename(columns={'Goalie Stats.16': 'GSAA'})
df = df.rename(columns={'Scoring': 'Goals'})
df = df.rename(columns={'Scoring.1': 'Assists'})
df = df.rename(columns={'Scoring.2': 'Points'})
df = df.rename(columns={'Scoring.3': 'Penalty Minutes'})
df = df.drop(columns=['-additional'])

goalies = df.to_dict(orient='records')

numeric_stats = []

for key in goalies[0]:
    if key != "Player":
        for goalie in goalies:
            if isinstance(goalie[key], (int, float)):
                numeric_stats.append(key)
                break

```

```python
name = input("Enter a goalie's name: ")

found = False
for goalie in goalies:
  if goalie['Player'].lower() == name.lower():
    found = True
    print("Goalie Stats:")
    for key in goalie:
      if key != 'Player':
        print(f"{key}: {goalie[key]}")
    break

if not found:
  print("Goalie not found.")
```

```python
import math

excluded_stats = ["Player", "From", "To", "Years Played"]
numeric_stats = []

for key in goalies[0]:
    if key not in excluded_stats:
        for goalie in goalies:
            try:
                value = float(goalie[key])
                if not math.isnan(value):
                    numeric_stats.append(key)
                    break
            except:
                continue

print("Stats you can search by:")
for stat in numeric_stats:
    print("-", stat)

chosen_stat = input("\nEnter the stat you want to find the best goalie for: ")

if chosen_stat not in numeric_stats:
    print("That stat is not valid. Please choose one from the list.")
else:
    lower_is_better = ["Goals Against", "Losses", "Goals Against Avg", "Really Bad Starts", "Penalty Minutes", "GA % -"]

    best_goalie = None
    best_value = None

    for goalie in goalies:
        value = goalie.get(chosen_stat)

        try:
            value = float(value)
            if math.isnan(value):
                continue
        except:
            continue

        if best_goalie is None:
            best_goalie = goalie
            best_value = value
        else:
            if chosen_stat in lower_is_better:
                if value < best_value:
                    best_goalie = goalie
                    best_value = value
            else:
                if value > best_value:
                    best_goalie = goalie
                    best_value = value

    if best_goalie:
        print(f"Best goalie based on {chosen_stat}:")
        print(f"{best_goalie['Player']} with {best_goalie[chosen_stat]}")
    else:
        print("Could not find a valid value for that stat.")
```
